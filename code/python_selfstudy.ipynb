## ۲. متغیرها و انواع داده
متغیر مثل یک **برچسب** است که روی یک مقدار می‌چسبانیم. با علامت `=` مقدار را در متغیر می‌گذاریم.

انواع داده‌ای که بیشتر استفاده می‌شوند:
- `int`: عدد صحیح، مثل `78`
- `float`: عدد اعشاری، مثل `36.8`
- `str`: متن، مثل `"Ali"`
- `bool`: درست یا غلط، یعنی `True` یا `False`

</div>

```python
heart_rate = 78          # int
temperature = 36.8       # float
name = "Ali"             # str
is_alarm = False         # bool

print(heart_rate, temperature, name, is_alarm)
print(type(heart_rate), type(temperature), type(name), type(is_alarm))
```

<div dir="rtl" style="text-align:right">

**امتحان کنید:** مقدار `heart_rate` را عوض کنید و سلول را دوباره اجرا کنید.

</div>

<div dir="rtl" style="text-align:right">

## ۳. محاسبات و متن
عملگرهای پایه: `+` جمع، `-` تفریق، `*` ضرب، `/` تقسیم، `**` توان.

برای ساختن متن با مقدار متغیرها از «f-string» استفاده می‌کنیم: قبل از نقل‌قول `f` می‌گذاریم و متغیر را داخل `{}` می‌نویسیم.

</div>

```python
weight = 70      # کیلوگرم
height = 1.75    # متر

bmi = weight / height ** 2
print(bmi)
print(f"شاخص توده بدنی: {bmi:.1f}")   # .1f یعنی یک رقم اعشار
```

<div dir="rtl" style="text-align:right">

## ۴. شرط‌ها (`if`)
با `if` برنامه بر اساس یک شرط تصمیم می‌گیرد. دقت کنید:
- بعد از `if`، `elif` و `else` علامت `:` می‌آید.
- خط‌های داخل شرط باید **تورفتگی** (۴ فاصله) داشته باشند.
- برای مقایسه از `>`، `<`، `>=`، `<=`، `==` (مساوی) و `!=` (نامساوی) استفاده می‌کنیم.

> عددهای آستانه در مثال‌ها فقط برای تمرین شرط‌ها هستند و **قاعده تشخیصی یا توصیه پزشکی نیستند**.

</div>

```python
heart_rate = 110

if heart_rate > 100:
    print("بالاتر از آستانه نمونه")
elif heart_rate < 50:
    print("پایین‌تر از آستانه نمونه")
else:
    print("در بازه نمونه")
```

<div dir="rtl" style="text-align:right">

**امتحان کنید:** `heart_rate` را یک بار `40` و یک بار `75` بگذارید و خروجی را ببینید.

</div>

<div dir="rtl" style="text-align:right">

## ۵. لیست‌ها
لیست مجموعه‌ای مرتب از مقدارهاست و با `[]` ساخته می‌شود. عضوها از **صفر** شماره‌گذاری می‌شوند.

</div>

```python
rates = [72, 80, 95, 110, 65]

print(rates)
print(rates[0])        # اولین عضو
print(rates[-1])       # آخرین عضو
print(len(rates))      # تعداد اعضا
print(max(rates), min(rates), sum(rates) / len(rates))

rates.append(88)       # افزودن یک عضو
print(rates)
```

<div dir="rtl" style="text-align:right">

## ۶. حلقه (`for`)
حلقه `for` یک کار را برای هر عضو لیست تکرار می‌کند. باز هم تورفتگی مهم است.

</div>

```python
rates = [72, 80, 95, 110, 65]

for r in rates:
    print("ضربان:", r)
```

```python
# شمارش مقدارهایی که از ۹۰ بیشترند
count = 0
for r in rates:
    if r > 90:
        count = count + 1

print("تعداد:", count)
```

<div dir="rtl" style="text-align:right">

## ۷. دیکشنری
دیکشنری اطلاعات را به صورت **«کلید: مقدار»** نگه می‌دارد و با `{}` ساخته می‌شود. با کلید به مقدار می‌رسیم.

</div>

```python
patient_info = {"name": "Ali", "age": 45, "hr": 78}

print(patient_info["age"])

patient_info["hr"] = 90          # تغییر یک مقدار
patient_info["spo2"] = 97        # افزودن کلید جدید
print(patient_info)
```

<div dir="rtl" style="text-align:right">

اگر کلیدی که وجود ندارد را بخواهید، خطای `KeyError` می‌گیرید. امتحان کنید: `patient_info["weight"]`

</div>

## ۸. تابع
تابع تکه‌کدی با یک نام است که می‌توانیم چند بار از آن استفاده کنیم. با `def` تعریف می‌شود، ورودی می‌گیرد و با `return` نتیجه را برمی‌گرداند.

</div>

```python
def average(values):
    return sum(values) / len(values)

print(average([72, 80, 95]))
print(average([60, 70]))
```

```python
def classify_hr(hr):
    if hr > 100:
        return "بالاتر از آستانه نمونه"
    elif hr < 50:
        return "پایین‌تر از آستانه نمونه"
    return "در بازه نمونه"

print(classify_hr(110))
print(classify_hr(70))
```

## ۹. کتابخانه‌ها: NumPy و Pandas
کتابخانه‌ها مجموعه‌ای از ابزارهای آماده‌اند. با `import` آن‌ها را وارد می‌کنیم.
- **NumPy:** محاسبات روی آرایه‌های عددی
- **Pandas:** کار با داده‌های جدولی (مثل اکسل)

</div>

```python
import numpy as np

x = np.array([72, 80, 95, 110, 65])
print(x.mean())     # میانگین
print(x.std())      # انحراف معیار
print(x.max())      # بیشینه
```

```python
import pandas as pd

# یک جدول کوچک و ساختگی
patients = pd.DataFrame({
    "name":  ["A", "B", "C", "D", "E"],
    "age":   [34, 51, 29, 62, 45],
    "hr":    [72, 88, 65, 95, 70],
    "group": ["control", "case", "control", "case", "control"],
})

patients
```

```python
print(patients.shape)                  # (تعداد ردیف، تعداد ستون)
print(patients["age"].mean())          # میانگین یک ستون
print(patients[patients["age"] > 40])  # انتخاب ردیف‌هایی که سن بیش از ۴۰ دارند
```

```python
patients.describe()    # آمار توصیفی ستون‌های عددی
```

```python
# میانگین ضربان به تفکیک گروه
patients.groupby("group")["hr"].mean()
```

```python
# مقدار گمشده: np.nan یعنی «اندازه‌گیری نشده»
patients_missing = patients.copy()
patients_missing.loc[2, "hr"] = np.nan

print(patients_missing["hr"].isna().sum())   # تعداد مقدارهای گمشده
```

## ۱۰. خواندن پیام خطا
خطا بخشی طبیعی از برنامه‌نویسی است. **آخرین خط** پیام خطا معمولاً مهم‌ترین است.

- `NameError`: متغیری که نوشتید تعریف نشده، یا سلول قبلی را اجرا نکرده‌اید.
- `KeyError`: نام کلید یا ستون اشتباه است.
- `IndentationError`: تورفتگی درست نیست.
- `TypeError`: نوع داده با کار مورد نظر سازگار نیست (مثلاً جمع کردن عدد با متن).
- `ModuleNotFoundError`: کتابخانه نصب یا وارد (import) نشده است.

**امتحان کنید:** سلول زیر را اجرا کنید و پیام خطا را بخوانید، سپس آن را درست کنید.

</div>

```python
print("ضربان:" + 72)
```

```python
# تمرین ۲
temps = [36.5, 37.2, 38.1, 36.9, 39.0]
count = 0
for t in temps:
    if t > 37.5:
        count = count + 1
print("تعداد:", count)
print("بیشینه:", max(temps))
```
