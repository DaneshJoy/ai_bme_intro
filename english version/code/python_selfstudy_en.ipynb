## 2. Variables and Data Types
A variable is like a **label** that we stick on a value. We put a value into a variable with the `=` sign.

The most commonly used data types:
- `int`: an integer, like `78`
- `float`: a decimal number, like `36.8`
- `str`: text, like `"Ali"`
- `bool`: true or false, that is `True` or `False`


```python
heart_rate = 78          # int
temperature = 36.8       # float
name = "Ali"             # str
is_alarm = False         # bool

print(heart_rate, temperature, name, is_alarm)
print(type(heart_rate), type(temperature), type(name), type(is_alarm))
```

## 3. Calculations and Text
Basic operators: `+` addition, `-` subtraction, `*` multiplication, `/` division, `**` power.

To build text that includes the values of variables, we use an "f-string": put an `f` before the quotation mark and write the variable inside `{}`.


```python
weight = 70      # kilograms
height = 1.75    # meters

bmi = weight / height ** 2
print(bmi)
print(f"Body mass index: {bmi:.1f}")   # .1f means one decimal place
```

## 4. Conditionals (`if`)
With `if`, the program makes a decision based on a condition. Pay attention to:
- A `:` comes after `if`, `elif`, and `else`.
- The lines inside a conditional must be **indented** (4 spaces).
- For comparison we use `>`, `<`, `>=`, `<=`, `==` (equal), and `!=` (not equal).

> The threshold numbers in the examples are only for practicing conditionals and are **not a diagnostic rule or medical recommendation**.


```python
heart_rate = 110

if heart_rate > 100:
    print("Above the sample threshold")
elif heart_rate < 50:
    print("Below the sample threshold")
else:
    print("Within the sample range")
```

**Try it:** Set `heart_rate` to `40` once and to `75` once, and look at the output.


## 5. Lists
A list is an ordered collection of values and is created with `[]`. Items are numbered starting from **zero**.


```python
rates = [72, 80, 95, 110, 65]

print(rates)
print(rates[0])        # first item
print(rates[-1])       # last item
print(len(rates))      # number of items
print(max(rates), min(rates), sum(rates) / len(rates))

rates.append(88)       # add an item
print(rates)
```

## 6. Loops (`for`)
A `for` loop repeats a task for each item of a list. Indentation matters here too.


```python
rates = [72, 80, 95, 110, 65]

for r in rates:
    print("Heart rate:", r)
```

```python
# Count the values greater than 90
count = 0
for r in rates:
    if r > 90:
        count = count + 1

print("Count:", count)
```

## 7. Dictionaries
A dictionary stores information as **"key: value"** pairs and is created with `{}`. We reach a value through its key.


```python
patient_info = {"name": "Ali", "age": 45, "hr": 78}

print(patient_info["age"])

patient_info["hr"] = 90          # change a value
patient_info["spo2"] = 97        # add a new key
print(patient_info)
```

If you ask for a key that does not exist, you get a `KeyError`. Try it: `patient_info["weight"]`


## 8. Functions
A function is a piece of code with a name that we can use several times. It is defined with `def`, takes inputs, and returns the result with `return`.


```python
def average(values):
    return sum(values) / len(values)

print(average([72, 80, 95]))
print(average([60, 70]))
```

```python
def classify_hr(hr):
    if hr > 100:
        return "Above the sample threshold"
    elif hr < 50:
        return "Below the sample threshold"
    return "Within the sample range"

print(classify_hr(110))
print(classify_hr(70))
```

## 9. Libraries: NumPy and Pandas
Libraries are collections of ready-made tools. We bring them in with `import`.
- **NumPy:** calculations on numerical arrays
- **Pandas:** working with tabular data (like Excel)


```python
import numpy as np

x = np.array([72, 80, 95, 110, 65])
print(x.mean())     # mean
print(x.std())      # standard deviation
print(x.max())      # maximum
```

```python
import pandas as pd

# A small, synthetic table
patients = pd.DataFrame({
    "name":  ["A", "B", "C", "D", "E"],
    "age":   [34, 51, 29, 62, 45],
    "hr":    [72, 88, 65, 95, 70],
    "group": ["control", "case", "control", "case", "control"],
})

patients
```

```python
print(patients.shape)                  # (number of rows, number of columns)
print(patients["age"].mean())          # mean of one column
print(patients[patients["age"] > 40])  # select the rows where age is over 40
```

```python
patients.describe()    # descriptive statistics of the numeric columns
```

```python
# Mean heart rate by group
patients.groupby("group")["hr"].mean()
```

```python
# Missing value: np.nan means "not measured"
patients_missing = patients.copy()
patients_missing.loc[2, "hr"] = np.nan

print(patients_missing["hr"].isna().sum())   # number of missing values
```

## 10. Reading Error Messages
Errors are a natural part of programming. The **last line** of an error message is usually the most important.

- `NameError`: The variable you wrote is not defined, or you have not run the previous cell.
- `KeyError`: The key or column name is wrong.
- `IndentationError`: The indentation is not correct.
- `TypeError`: The data type is not compatible with the operation (for example, adding a number to text).
- `ModuleNotFoundError`: The library is not installed or has not been imported.
